```python
import numpy as np
import pandas as pd

idades = np.array([18, 21, 19, 22, 20])
print("1. Vetor de idades:", idades)

matriz = np.array([[10, 20, 30], [40, 50, 60]])
print("\n2. Matriz:\n", matriz)
print("   Shape da matriz:", matriz.shape)
print("   Elemento [1, 2]:", matriz[1, 2])

indices = ["Ana", "Bruno", "Carla", "Diego", "Eva"]
serie_idades = pd.Series(idades, index=indices)
print("\n3. Series de idades:\n", serie_idades)

idade_carla = serie_idades.loc["Carla"]
print(f"\n4. Idade da Carla (usando .loc): {idade_carla}")

maiores_21 = serie_idades[serie_idades >= 21]
print("\n5. Pessoas com 21 anos ou mais:\n", maiores_21)
```
